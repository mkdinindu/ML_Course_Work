```python
import pandas as pd

df = pd.read_csv("/content/parkinson_data/parkinsons_disease_dataset.csv")
df.head()
```

Now i'll preprocess the data set

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df['Sex'] = df['Sex'].map({"Male": 0, "Female": 1})
df['Family_History'] = df['Family_History'].map({"No": 0, "Yes": 1})
```

```python
df.head()
```

```python
df['Parkinsons_Status'].value_counts()
```

```python
X = df.drop(['Parkinsons_Status'], axis=1)
y = df['Parkinsons_Status']
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=82, stratify=y)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, classification_report

models = {
    "Logistic Regression": LogisticRegression(max_iter=100, class_weight='balanced'),
    "Random Forest": RandomForestClassifier(class_weight='balanced'),
    "SVM": SVC(class_weight='balanced')
}

for name, model in models.items():
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)
    print(f"\nModel: {name}")
    print("Accuracy:", accuracy_score(y_test, y_pred))
    print("Classification Report:\n", classification_report(y_test, y_pred))
```
