```python
import pandas as pd

#import kagglehub

# Download latest version
#path = kagglehub.dataset_download("miadul/parkinsons-disease-synthetic-dataset")

parkinsonsData = pd.read_csv("parkinsons_disease_dataset.csv")
parkinsonsData.head()
```

```python
y = parkinsonsData.Parkinsons_Status
```

```python
parkinsonsData['Sex'] = parkinsonsData['Sex'].map({"Male":0,"Female":1})
parkinsonsData['Family_History'] = parkinsonsData['Family_History'].map({"No":0,"Yes":1})
```

```python
features = ['Age','Sex','Tremor_Intensity','Voice_Pitch','Voice_Jitter','Voice_Shimmer','Bradykinesia_Score','Postural_Instability','Rigidity_Level','UPDRS_Score','Handwriting_Change','Walking_Difficulty','Family_History','MRI_Score']
X = parkinsonsData[features]
```

```python
# Import the train_test_split function
from sklearn.model_selection import train_test_split

train_X, val_X, train_y, val_y = train_test_split(X ,y ,random_state = 1)
```

```python
parkinson_detection_model = DecisionTreeRegressor(random_state = 1)

# Fit parkinson_detection_model with the training data.

parkinson_detection_model.fit(train_X,train_y)
```

```python
# Predict with all validation observations
val_predictions = parkinson_detection_model.predict(val_X)
```

```python
# print the top few validation predictions
print(val_predictions)
# print the top few actual results from validation data
print(val_y)
```

```python
from sklearn.metrics import mean_absolute_error
val_mae = mean_absolute_error(val_y,val_predictions)


print(val_mae)
```

```python
def get_mae(max_leaf_nodes,train_X,val_X,train_y,val_y):
    model = DecisionTreeRegressor(max_leaf_nodes = max_leaf_nodes, random_state= 0)
    model.fit(train_X, train_y)
    preds_val = model.predict(val_X)
    mae = mean_absolute_error(val_y, preds_val)
    return(mae)
```

```python
candidate_max_leaf_nodes = [5, 25, 50, 100, 250, 500]
meas = []
#find the ideal tree size from candidate_max_leaf_nodes
for max_leaf_nodes in candidate_max_leaf_nodes:
    my_mea = get_mae(max_leaf_nodes, train_X, val_X, train_y, val_y)
    meas.append(my_mea)
    print("max leaf nodes: ",max_leaf_nodes,"  MAE: ", my_mea)
    if min(meas) == my_mea:
        best_tree_size = max_leaf_nodes

# Store the best value of max_leaf_nodes (it will be either 5, 25, 50, 100, 250 or 500)
#best_tree_size = min(meas)
print(best_tree_size)
```

```python
candidate_max_leaf_nodes = [2, 3, 4, 5, 6,7]
meas = []
#find the ideal tree size from candidate_max_leaf_nodes
for max_leaf_nodes in candidate_max_leaf_nodes:
    my_mea = get_mae(max_leaf_nodes, train_X, val_X, train_y, val_y)
    meas.append(my_mea)
    print("max leaf nodes: ",max_leaf_nodes,"  MAE: ", my_mea)
    if min(meas) == my_mea:
        best_tree_size = max_leaf_nodes

# Store the best value of max_leaf_nodes (it will be either  2, 3, 4, 5, 6 or 7)
#best_tree_size = min(meas)
print(best_tree_size)
```

```python
# Fill in argument to make optimal size and uncomment
final_model = DecisionTreeRegressor(max_leaf_nodes=best_tree_size,random_state=1)

# fit the final model and uncomment the next two lines
final_model.fit(X,y)
```

```python
from sklearn.ensemble import RandomForestRegressor

#Define the model. Set random_state to 1
rf_model = RandomForestRegressor(random_state=1)

#fit model
rf_model.fit(train_X,train_y)
rf_val_predictions = rf_model.predict(val_X)

# Calculate the mean absolute error of Random Forest model on the validation data
rf_val_mae = mean_absolute_error(rf_val_predictions, val_y)

print("Validation MAE for Random Forest Model: {:}".format(rf_val_mae))
```

```python
# To improve accuracy, creating a new Random Forest model which will train on all training data
rf_model_on_full_data = RandomForestRegressor(random_state=1)

# fit rf_model_on_full_data on all data from the training data
rf_model_on_full_data.fit(X,y)
```

```python
# The list of columns is stored in a variable called features
val_X = val_X[features]
# make predictions which we will submit.
test_preds = rf_model_on_full_data.predict(val_X)
```

```python
# Calculate the mean absolute error of Random Forest model on the validation data
rf_val_mae = mean_absolute_error(test_preds, val_y)

print("Validation MAE for Random Forest Model: {:}".format(rf_val_mae))
```

```python
# Create an empty list to store the new, thresholded predictions
status_thresholded = []

# Loop through each prediction
for pred in test_preds:
    if pred >= 0.5:
        status_thresholded.append(1)
    else:
        status_thresholded.append(0)

# Create the output DataFrame with the thresholded predictions
output = pd.DataFrame({'Id': val_X.index,
                       'Status': status_thresholded})
output.to_csv('submission.csv', index=False)
```
